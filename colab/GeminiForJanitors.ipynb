```python
# @title Proxy Runner { display-mode: "form" }

commands = """
set -e

curl -LO https://github.com/cloudflare/cloudflared/releases/latest/download/cloudflared-linux-amd64
chmod +x cloudflared-linux-amd64
export GFJPROXY_CLOUDFLARED="$PWD/cloudflared-linux-amd64"

[ ! -d GeminiForJanitors ] && \
  git clone --depth=1 https://github.com/vu5eruz/GeminiForJanitors.git

cd GeminiForJanitors

export GFJPROXY_ADMIN="Google Colab"
export GFJPROXY_COOLDOWN=0
export GFJPROXY_DEVELOPMENT=yes
export GFJPROXY_PROCESS_TIMEOUT=300
export GFJPROXY_XUID_SECRET="Google Colab"
export PORT=5000

uv run --no-dev gunicorn -b 0.0.0.0:${PORT} -k gevent -w 1 -t ${GFJPROXY_PROCESS_TIMEOUT} "gfjproxy.app:create_app()"
"""

with open("run.sh", "w") as f:
    f.write(commands)

!bash ./run.sh
```
